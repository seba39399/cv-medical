```python
import os
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Ruta de ejemplo a un sujeto descargado
subject_path = r"C:\Users\juan_\Desktop\cv-medical\COBRE_Dataset\sub-10159\anat\sub-10159_T1w.nii.gz"

# 1. Cargar el archivo NIfTI con nibabel
nii_img = nib.load(subject_path)
mri_data = nii_img.get_fdata()

print(f"Dimensiones del volumen 3D del cerebro: {mri_data.shape}")
# Por lo general te dará algo como (256, 256, 176)

# 2. Extraer un corte (slice) central del cerebro en el eje axial
slice_idx = mri_data.shape[2] // 2
axial_slice = mri_data[:, :, slice_idx]
```

```python
# 3. Visualizar rápido para verificar que cargó bien
plt.imshow(axial_slice.T, cmap="gray", origin="lower")
plt.title(f"Corte Axial MRI - Sujeto 10159 (Slice {slice_idx})")
plt.axis("off")
plt.show()
```
